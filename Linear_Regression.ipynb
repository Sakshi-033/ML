```python
df = pd.get_dummies(df)
```

```python
x = df.drop('G3', axis=1)
y = df['G3']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_temp,y_train,y_temp = train_test_split(
    x,
    y,
    test_size= 0.4,
    random_state= 42
)
```

```python
x_validation,x_test,y_validation,y_test = train_test_split(
    x_temp,
    y_temp,
    test_size= 0.5,
    random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
```

```python
x_validation_scaled = scaler.fit_transform(x_validation)
```

```python
x_test_scaled = scaler.fit_transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics  import mean_squared_error
```

```python
model = LinearRegression()
```

```python
model.fit(x_train_scaled,y_train)
prediction = model.predict(x_train_scaled)
train_mse = mean_squared_error(y_train,prediction)
train_rmse = train_mse**0.5
```

```python
#validation 

prediction = model.predict(x_validation_scaled)
validation_mse = mean_squared_error(y_validation,prediction)
validation_rmse = validation_mse**0.5
```

```python
#testing

prediction = model.predict(x_test_scaled)
test_mse = mean_squared_error(y_test,prediction)
test_rmse = test_mse**0.5
```

```python
print(
    "==========================================\n",
    "TRAINNING ================================\n",
    "Training mse = ",train_mse,"\n",
    "Training rmse = ",train_rmse,"\n",
    
    "===========================================\n",
    "VALIDATION ================================\n",
    "Validation mse = ",validation_mse,"\n",
    "Validation rmse = ",validation_rmse,"\n",
    
    "============================================\n",
    "TESTING=====================================\n",
    "Testing mse = ", test_mse,"\n",
    "Testing rmse = ", test_rmse,"\n",
        
    
)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

alphas = [0.001,0.01, 0.1, 1, 10, 100, 1000]
for alpha in alphas:
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(x_train_scaled,y_train)
    
    prediction = ridge_model.predict(x_train_scaled)
    train_mse = mean_squared_error(y_train,prediction)
    train_rmse = train_mse**0.5
    
    prediction2 = ridge_model.predict(x_validation_scaled)
    validation_mse = mean_squared_error(y_validation,prediction2)
    validation_rmse = validation_mse**0.5
    
    prediction3 = ridge_model.predict(x_test_scaled)
    test_mse = mean_squared_error(y_test,prediction3)
    test_rmse = test_mse**0.5
    
    print("Alphas = ",alpha,"train_mse = ", train_mse,"train_rmse = ", train_rmse,"\n",
          "alphas = ",alpha,"validation_mse =",validation_mse, "validation_rmse = ",validation_rmse,"\n",
          "alphas = ",alpha, "test_mse = ",test_mse,"test_rmse =",test_rmse,"\n"
          
          )
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

alphas = [0.001, 0.01, 0.1, 1, 10, 100, 1000]

for alpha in alphas:
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(x_train_scaled, y_train)

    prediction = ridge_model.predict(x_train_scaled)
    train_mse = mean_squared_error(y_train, prediction)
    train_rmse = train_mse ** 0.5

    prediction2 = ridge_model.predict(x_validation_scaled)
    validation_mse = mean_squared_error(y_validation, prediction2)
    validation_rmse = validation_mse ** 0.5

    print(
        "alpha =", alpha,
        "train_rmse =", train_rmse,
        "validation_rmse =", validation_rmse
    )
```
